```python
df_country=pd.read_excel('Country-Code.xlsx')
df_country.head()
```

```python
df.columns
```

```python
final_df=pd.merge(df,df_country,on='Country Code',how='left')
```

```python
final_df.head(2)
```

```python
final_df.dtypes
```

```python
final_df.Country.value_counts()
```

```python
country_names=final_df.Country.value_counts().index
country_names
```

```python
country_values=final_df.Country.value_counts().values
country_values
```

```python
# Top 3 countries using Zomato
plt.pie(country_values[:3],labels=country_names[:3],autopct='%1.2f%%')
```

### Observation 1

-->Highest number of transactions are recorded in India followed by United States and United Kingdom

```python
final_df.columns
```

```python
ratings=final_df.groupby(['Aggregate rating','Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
```

```python
ratings
```

### Observation 2

1. Rating ranges from 4.5 to 4.9 indicates ----> Excellent
2. Rating ranges from 4.0 to 4.4 indicates ----> Very Good
3. Rating ranges from 3.5 to 3.9 indicates ----> Good
4. Rating ranges from 2.5 to 3.4 indicates ----> Average
5. Rating ranges from 1.8 to 2.4 indicates ----> Poor

```python
import matplotlib
matplotlib.rcParams['figure.figsize']=(12,6)
sns.barplot(x='Aggregate rating',y='Rating Count',data=ratings)
```

```python
sns.barplot(x='Aggregate rating',y='Rating Count',hue='Rating color',data=ratings,palette=['blue','red','orange','yellow','green','green'])
```

### Observation 3

1. Not rated has the highest count
2. Highest number of ratings are between 2.5 to 3.4

```python
# count plot
sns.countplot(x='Rating color',data=ratings,palette=['blue','red','orange','yellow','green','green'])
```
